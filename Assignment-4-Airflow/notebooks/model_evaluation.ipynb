# Model Evaluation

This notebook evaluates the trained Logistic Regression model using the preprocessed test data.

### Objectives
- Load the trained model
- Load the test data
- Generate predictions
- Calculate model accuracy
- Generate an evaluation report
- Save the report for the automated workflow

```python
import pandas as pd
import joblib
from sklearn.metrics import accuracy_score, classification_report

# Load test data
X_test = pd.read_csv("../data/X_test.csv")
y_test = pd.read_csv("../data/y_test.csv").squeeze()

# Load trained model
model = joblib.load("../models/trained_model.pkl")

# Generate predictions
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Generate classification report
report = classification_report(y_test, y_pred)

# Display results
print("Model evaluation completed successfully.")
print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(report)

# Save evaluation report
report_path = "../output/training_report.txt"

with open(report_path, "w") as file:
    file.write("ML MODEL TRAINING REPORT\n")
    file.write("========================\n")
    file.write("Model: Logistic Regression\n")
    file.write(f"Accuracy: {accuracy:.4f}\n\n")
    file.write("Classification Report:\n")
    file.write(report)

print(f"Report saved to: {report_path}")
```
