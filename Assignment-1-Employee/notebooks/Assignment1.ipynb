```python
import pandas as pd
```

# Version 1 - Initial Employee Salary Dataset

## Objective
Create the initial employee salary dataset containing:
- Employee_ID
- Employee_Name
- Department
- Salary

This dataset will be tracked using DVC for version control.

```python
employees = {
    "Employee_ID": [201,202,203,204,205,206,207,208,209,210],
    "Employee_Name": [
        "Aarav Sharma","Diya Patel","Kabir Singh","Meera Nair","Rohan Verma",
        "Sneha Iyer","Vivaan Gupta","Ananya Rao","Aryan Mehta","Ishita Kapoor"
    ],
    "Department": [
        "HR","Finance","IT","Marketing","Sales",
        "Operations","IT","Finance","HR","Marketing",
    ],
    "Salary": [
        45000, 60000, 55000, 52000, 48000,
        50000, 65000, 58000, 47000, 54000
    ]
}
```

```python
df = pd.DataFrame(employees)
df
```

```python
df.to_csv("../data/employees.csv", index=False)
print("Version 1 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/employees.csv")
df_loaded
```

# Version 2 - Salary Increment

## Objective

Increase the salary of every employee by **10%** and create a new version of the dataset using DVC.

```python
df["Salary"] = (df["Salary"] * 1.10).round().astype(int)

df
```

```python
df.to_csv("../data/employees.csv", index=False)
print("Version 2 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/employees.csv")
df_loaded
```

# Version 3 - Add Years of Experience

## Objective

Add a new column named **Years_of_Experience** to demonstrate dataset schema evolution using DVC.

```python
import random
random.seed(42)
df["Years_of_Experience"] = [
    random.randint(1, 15)
    for _ in range(len(df))
]
```

```python
df.to_csv("../data/employees.csv", index=False)
print("Version 3 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/employees.csv")
df_loaded
```

# Version 5 - Add New Employee Records

## Objective

Add three new employee records to the dataset. This demonstrates dataset growth by increasing the number of records while preserving the existing schema.

Changes Made:
- Added three new employee records.
- Saved the updated dataset.
- Tracked the dataset using DVC.

```python
new_employees = pd.DataFrame({
    "Employee_ID": [211, 212, 213],
    "Employee_Name": [
        "Rahul Joshi",
        "Priya Deshmukh",
        "Karan Malhotra"
    ],
    "Department": [
        "IT",
        "Finance",
        "Marketing"
    ],
    "Salary": [
        62000,
        59000,
        61000
    ],
    "Years_of_Experience": [
        4,
        6,
        5
    ],
    "Performance_Rating": [
        "Good",
        "Excellent",
        "Very Good"
    ]
})

df = pd.concat([df, new_employees], ignore_index=True)
df
```

```python
df.to_csv("../data/employees.csv", index=False)
print("Version 5 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/employees.csv")
df_loaded
```

# Version 6 - Update Department Names

## Objective

Modify department names to reflect organizational restructuring. This demonstrates updating categorical values within the dataset.

Changes Made:
- Renamed department values.
- Saved the updated dataset.
- Tracked the dataset using DVC.

```python
df["Department"] = df["Department"].replace({
    "HR": "Human Resources",
    "IT": "Engineering",
    "Finance": "Accounts & Finance",
    "Marketing": "Business Development",
    "Sales": "Sales & Marketing",
    "Operations": "Business Operations"
})
df
```

```python
df.to_csv("../data/employees.csv", index=False)
print("Version 6 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/employees.csv")
df_loaded
```

- Version 3: Added Years_of_Experience column
