# Assignment 6 - House Price Prediction
## 06. Docker Containerisation and Testing

## 1. Create Dockerfile

````python
from pathlib import Path
PROJECT_ROOT=Path("..").resolve()

dockerfile='''FROM python:3.12-slim
WORKDIR /app
ENV PYTHONDONTWRITEBYTECODE=1
ENV PYTHONUNBUFFERED=1
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt
COPY app ./app
COPY model ./model
EXPOSE 8000
CMD ["uvicorn","app.main:app","--host","0.0.0.0","--port","8000"]
'''
(PROJECT_ROOT/"Dockerfile").write_text(dockerfile,encoding="utf-8")
print("Dockerfile created.")
````

## 2. Create .dockerignore

````python
dockerignore='''__pycache__/
*.pyc
.ipynb_checkpoints/
.git/
.venv/
venv/
*.ipynb
notebooks/
data/
'''
(PROJECT_ROOT/".dockerignore").write_text(dockerignore,encoding="utf-8")
print(".dockerignore created.")
````

## 3. Build Docker Image
Run from the project root:

```bash
docker build -t house-price-api:1.0 .
```

## 4. Run Container

```bash
docker run --name house-price-api -p 8000:8000 house-price-api:1.0
```

Then open `http://localhost:8000/docs`.

## 5. Test Health Endpoint

```bash
curl http://localhost:8000/health
```

Expected response contains `"status":"healthy"` and `"model_loaded":true`.

## 6. Stop Container

```bash
docker stop house-price-api
docker rm house-price-api
```

## Conclusion
The FastAPI ML service and persisted model pipeline are packaged into a lightweight Python slim Docker image exposing port 8000.
