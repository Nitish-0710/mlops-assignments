# Practice Assignment - Dataset Versioning using DVC

## Objective

The objective of this practice assignment is to understand the workflow of **Dataset Versioning using Data Version Control (DVC)**.

A custom student dataset is created and versioned using DVC. Two versions of the dataset are maintained to demonstrate dataset growth through the addition of new records.

# Version 1 - Initial Student Dataset

## Objective

Create the initial student dataset containing:

- Student_ID
- Name
- Age
- Branch
- CGPA

The dataset initially contains **10 student records** and will be tracked using DVC.

```python
import pandas as pd
```

```python
students = {
    "Student_ID": [101,102,103,104,105,106,107,108,109,110],
    "Name": [
        "Aarav","Diya","Kabir","Meera","Rohan",
        "Sneha","Vivaan","Ananya","Aryan","Ishita"
    ],
    "Age": [
        20,21,19,22,20,
        21,23,20,22,19
    ],
    "Branch": [
        "CSE","AI","IT","CSE","AI",
        "ECE","IT","AI","CSE","ECE"
    ],
    "CGPA": [
        8.5,9.1,7.8,8.9,9.3,
        8.0,7.5,9.0,8.4,8.8
    ]
}
```

```python
df = pd.DataFrame(students)
df
```

```python
df.to_csv("../data/students.csv", index=False)
print("Version 1 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/students.csv")
df_loaded
```

# Version 2 - Add New Student Records

## Objective

Add **three new student records** to the existing dataset.

Changes Made:

- Added three new students.
- Increased dataset size from **10** to **13** records.
- Saved the updated dataset.
- Tracked the updated dataset using DVC.

```python
new_students = pd.DataFrame({
    "Student_ID": [111,112,113],
    "Name": [
        "Rahul",
        "Priya",
        "Karan"
    ],
    "Age": [
        21,
        20,
        22
    ],
    "Branch": [
        "AI",
        "CSE",
        "IT"
    ],
    "CGPA": [
        8.7,
        9.2,
        8.1
    ]
})
```

```python
df = pd.concat([df, new_students], ignore_index=True)
df
```

```python
df.to_csv("../data/students.csv", index=False)
print("Version 2 dataset saved successfully.")
```

```python
df_loaded = pd.read_csv("../data/students.csv")
df_loaded
```

# Conclusion

In this practice assignment, dataset versioning was implemented using **Data Version Control (DVC)**.

The dataset evolved through two versions:

- **Version 1:** Created the initial student dataset containing **10 records**.
- **Version 2:** Added **three new student records**, increasing the dataset size to **13 records**.

Each version was tracked using **DVC** and committed using **Git**, demonstrating how dataset versions can be managed efficiently in Machine Learning projects.
